## Load profile shape and value distribution

The plot shows the relative consumption value (0-1) per quarter-hour, across the
full year 2023 (35,040 points = 365 days x 96 quarters/day). Each sharp spike is
one day; peak height varies slightly across the year. `WAARDE` is an average
across hundreds of offices (their value at that specific moment, not their
individual peaks), so it's less extreme than any single real office would be.
Max = 0.9, min = 0.31 — confirms this is a normalized ("renorm") profile, not
absolute kWh.

**Density check:** the line plot visually looked "denser" around ~0.5-0.6 than
near the extremes. Verified with a histogram (counts occurrences directly,
independent of how the line plot renders overlapping points) — confirmed: values
around 0.5-0.6 occur most often. Likely explanation: the daily curve spends only
a short time at the exact peak (~0.9) or exact trough (~0.3), but passes through
the mid-range twice a day (rising in the morning, falling in the evening) — not
a data issue, just a natural consequence of the daily shape.

```python
probe = get_load_profile(peak_kwh_per_quarter=1.0)
profile_sum = probe.sum()

building_area_m2 = 590
target_annual_kwh = building_area_m2 * 115
peak_kwh_per_quarter = target_annual_kwh / profile_sum

load = get_load_profile(peak_kwh_per_quarter=peak_kwh_per_quarter)
peak_kw = load.max() / 0.25

print(f"peak_kwh_per_quarter = {peak_kwh_per_quarter:.2f}")
print(f"Piek: {peak_kw:.1f} kW")
```
